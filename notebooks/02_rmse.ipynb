```python
#RMSE 오차 코드 실습
#y = 1.96x + 5.5
# weight와 bias를 알고 있을 때 오차 구하기
#np.sqrt():제곱근 계산

import numpy as np

y_true = [21, 49, 68, 80] #실제값
y_pred = [25.1, 44.7, 64.3, 83.9] #예측값

#RMSE 계산 함수
def rmse(y_true, y_pred):
    #list끼리는 -빼기연산 못함
    #unsupported operand type(s) for -: 'list' and 'list'
    mse = np.mean((np.array(y_true) - np.array(y_pred)) ** 2)
    #numpy배열로 바꿔야 각 요소끼리 -연산가능(백터연산)
    return np.sqrt(mse)

print(f"RMSE: {rmse(y_true, y_pred)}")
#RMSE: 4.006245124802027

#list는 백터/행렬 연산 불가, for문 일일이 돌리면 속도 저하
#np.array():행렬 연산 가능, list 또는 튜플을 아규먼트로 전달
# 옵션 파라미터 : dtype 지정 가능
# 예 dtype=float라고 해서 호출하면 넘파이 배열안의 요소를 
# float타입 실수형 데이터로 변환해서 반환함
# np.array()의 반환값 numpy.ndarray 넘파이배열 차원객체리턴
#for문없이 빠르고 편하게 계산하려고 넘파이배열로 변환해서 사용하는 것
# np.array(튜플또는리스트): return 넘파이배열(ndarray)
# numpy배열로 변환하는 함수
# 파이썬의 튜플이나 list를 수학 계산용 배열 객체로 바꿈
# AI 데이터는 대부분 숫자 백터/행렬
# 각각의 배열 요소들간의 계산 -> 백터연산
# 파이썬의 list자료구조로는 불편 for문 돌려야함
# numpy배열 빠르게 가능
#결과 출력
```
