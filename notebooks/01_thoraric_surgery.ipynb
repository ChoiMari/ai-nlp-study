```python
#텐서플로우 설치확인
import tensorflow as tf
print(tf.__version__)#2.21.0
```

```python
#맛보기_폐암 수술 예측
#Data - ThoraricSurgery.csv

#딥러닝 구동하는 데 필요한 케라스 함수를 불러온다
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
#필요한 라이브러리를 불러온다
import numpy as np
import tensorflow as tf

#실행할 때마다 같은 결과를 출력하기 위해 설정
np.random.seed(3)
tf.random.set_seed(3)

#준비된 수술 환자 데이터를 불러옴
Data_set = np.loadtxt("../data/ThoraricSurgery.csv", delimiter=',')

#환자의 기록과 수술 결과를 X, Y로 구분하여 저장한다
X = Data_set[:, 0:17]
Y = Data_set[:, 17]

#딥러닝 구조 결정 - 모델을 설정하고 실행하는 부분
model = Sequential()
model.add(Input(shape=(17,))) #2.X 후반에 변경된 부분, Input레이어 사용
model.add(Dense(30, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
#딥러닝 실행
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X,Y,epochs=100,batch_size=10)
print("\n Accuracy: %.4f" % (model.evaluate(X,Y)[1]))

#Accuracy: 0.8574
```

```python
#-----------------------------------------------------
# 코드 해석
# 폐암 수술 환자의 데이터를 보고
# 수술 후 생존/사망 여부를 예측하는 이진 분류 딥러닝 코드

#데이터 입력 -> 모델 예측 -> 실제 정답과 비교(loss)
# -> 틀린 정도 계산 -> 파라미터(가중치, 편향) 수정 -> 반복 학습

#딥러닝 구동하는 데 필요한 케라스 함수를 불러온다
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
# Sequential: 층(layer)을 순서대로 쌓는 모델
# Dense: 완전 연결층(Fully Connected Layer), 모든 뉴런 서로 연결된 층
# 입력값 하나하나가 모든 뉴런에 연결 됨
# 예시
# 입력 17개
# ↓↓↓↓↓↓↓↓↓
# 은닉 뉴런 30개
# Input: 입력 데이터 형태 정의

#필요한 라이브러리를 불러온다
import numpy as np #행렬 계산용
import tensorflow as tf #딥러닝 엔진

#시드 설정 : 실행할 때마다 같은 결과를 출력하기 위해 설정
# 매우 중요,, 딥러닝은 랜덤성이 있다, 그래서 실행할 때마다
# 결과가 달라질 수 있어서,, seed를 고정해야 재현성 확보됨
#랜덤값인데 같은 랜덤값이 나오도록 고정
np.random.seed(3)
tf.random.set_seed(3)

#1. 데이터 로드
#준비된 수술 환자 데이터를 불러옴
Data_set = np.loadtxt("../data/ThoraricSurgery.csv", delimiter=',')
#csv파일 읽는 부분, delimiter 구분자 ,쉼표
#불러온 데이터에서 X, Y 분리함
#환자의 기록과 수술 결과를 X, Y로 구분하여 저장한다
X = Data_set[:, 0:17] #입력데이터: 모든행, 0~16열
Y = Data_set[:, 17]#정답데이터: 모든행, 17열
# X :입력, feature, 대상의 특징, 검사결과
# Y: 정답, 클래스, label
#.csv파일에서 0~16열까지가 각 환자의 검사결과
#17열이 사망 또는 생존 데이터이므로 저렇게 파싱한 것
# numpy 배열에서 열을 몇 개 선택하느냐에 따라
# 1차원배열(백터), 2차원 배열(행렬)이 결정됨
#X는 2차원 배열
#[
# [65, 1, 3, 0, ...],   # 환자1
# [72, 0, 2, 1, ...],   # 환자2
#]
#Y는 1차원 배열
#[1,0,1,1,0,1]
# numpy는 정수로 열 하나만 선택하면 자동으로 차원을 축소한다.
#만약, Y도 2차원으로 받고 싶으면
# Y = Data_set[:, [17]] 또는 Y = Data_set[:, 17:18]
print(X.shape)
print(Y.shape)

#2. 신경망 구조 결정 - 모델을 설정하고 실행하는 부분
model = Sequential()#빈 신경망 생성

#입력 데이터 구조 정의
model.add(Input(shape=(17,))) #2.X 후반에 변경된 부분, Input레이어 사용
#Input Layer 입력층: 입력데이터는 17개 feature을 가진다라는 뜻
#0~16 총 17개
#shape=(17,) : 길이가 17인 1차원 벡터
# 뒤에 , 붙는건 튜블 문법 때문
#대상 1개의 데이터 구조를 정의하는 것.
#여기서는 환자 1명의 데이터 입력 구조를 정의
#[나이, 흡연여부, 폐기능, ..., 기타] 이렇게 feature 17개 가진 백터 1개
#Input()함수 호출하면서 shape파라미터 변수에 (17,)아규먼트값 넣어준것
# Input()의 역할: 앞으로 들어올 데이터 구조 설명, 앞으로 들어올 데이터는 17개 feature가진 백터라고
#알려주는 것, 실제 데이터는 model.fit에서 들어감
model.add(Dense(30, activation='relu'))
#은닉층: 은닉 뉴런 30개 생성
#뉴런 하나는 y=f(w1x1+w2x2+...+b)
model.add(Dense(1, activation='sigmoid'))
#출력층: Dense(1 출력값 1개 이진 분류라서,,
#activation='sigmoid': 출력을 0~1사이의 확률데이터로 변환

#3. 딥러닝 실행 전 학습 방법 정의
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
#loss는 모델이 얼마나 틀렸는지 계산
#정답과 예측을 비교할 때 이 공식을 써라
#optimizer 파라미터(가중치, 편향) 최적화 알고리즘
#metrics=['accuracy']는 학습 중 어떤 평가 지표를 출력할 건지 설정
# 사람이 보기 위한 성능 지표
# metrics는 학습에 영향 안줌
#"평가할 때 accuracy도 계산해라", 안쓰면 loss만 계산함
#loss,옵티마이저 함수설정, metrics계산 규칙 모델 내부에 등록하는것

#4. 학습
model.fit(X,Y,epochs=100,batch_size=10)
#epochs=100 전체 데이터셋을 100번 반복학습
#batch_size=10 한 번에 10명씩 학습
# 470명 전체를 10명 단위로 나눠서 학습
#전체를 한번에 하면 메모리 많이 먹고 느려서,,
#설정 안하면 keras기본값 32

#epoch:전체 데이터 몇 바퀴 돌릴까
#batch:몇 개씩 끊어서 학습할까
# 470 전체 데이터를 (470 나누기 batch_size = 47번)
# 에포트 1번당 47번(배치사이즈10개씩 묶어서 돌리니까)돌림
# 100번 돌리면 100 * 47 = 총 4700번 반복
#10명(배치사이즈만큼) 입력 -> loss계산 -> gradient 계산 -> 옵티마이저알고리즘으로
# weight/bias 수정 -> 다음 10명 반복

#X입력 -> 예측값 생성 -> loss계산 -> 역전파 -> gradient계산 -> 옵티마이저 adam알고리즘으로
# -> 파라미터(weight, bias) 업데이트 -> 반복

#5. 평가
print("\n Accuracy: %.4f" % (model.evaluate(X,Y)[1]))
# 모델 성능 평가
# model.evaluate(X, Y)의 반환값에서 [1] 
#인덱스 1번째 값을 꺼내 출력
# model.evaluate(X, Y) : X넣어서 예측, Y정답과 비교
# loss계산, metrixs계산 -> 결과 반환
# 반환 값은 compile코드에 metrics=['accuracy']추가해서
#evaluate는 loss + accuracy를 반환
#반환리스트의 인덱스 0 : loss, 1: accuracy
#loss, accuracy = model.evaluate(X,Y)
#print(accuracy)

#evaluate 내부
#X넣음 -> 예측 생성 -> Y와 비교 -> loss계산 -> metrics계산 -> 결과반환
#왜 accuracy 계산 가능한가? compile에서 metrics=['accuracy']등록해서
#evaluate는 compile때 저장된 설정을 읽어와서 자동계산
#model.evaluate(X,Y) : 현재 weight상태로 X를 예측했을 때 Y와 얼마나 맞는지?
# fit()은 학습하며 파라미터 수정, evaluate()는 성능 측정만 함
#현재 모델로 x를 예측한 뒤 Y와 비교해서 compile 때 등록한 loss/metrics를 계산하는 함수
# 이 코드의 문제점:(강의 초반이라서 일부러 그런듯)
# model.fit(X, Y)와 model.evaluate(X, Y)에
# 동일한 데이터를 사용하고 있음

# 즉, 학습 데이터와 평가 데이터가 분리되지 않음

# 이 경우 모델이 데이터를 "일반화"한 것이 아니라
# 단순 암기(overfitting)했더라도
# 높은 accuracy가 나올 수 있음

# 따라서 train / validation / test 데이터를
# 분리해서 평가해야 함
#Accuracy: 0.8574


```
