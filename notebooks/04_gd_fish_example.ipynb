```python
#경사하강법 적용 코드 - 예제
# 물고기 사육주수 data를 통해 
# 경사 하강법을 구현하고 결과 확인해보기

# 기대 결과 w:0.1756, b:-0.03392
# y = 0.1756x - 0.03392

import matplotlib.pyplot as plt

#물고기 사육주수에 따른 물고기 크기
data = [[1,0.2],[2,0.3],[3,0.5],[4,0.6],[5,0.9],[6,0.95],[7,1.1],[8,1.5]]

x = [i[0] for i in data] #사육기간
#[1,2,3,4,5,6,7,8]
y = [i[1] for i in data] #크기
#[0.2,0.3,0.5,0.6,0.9,0.95,1.1,1.5]

#확인
print(x,y)
plt.figure(figsize=(8,5))
plt.scatter(x,y)
plt.show()

import numpy as np

#백터 계산 위해 넘파이 배열(ndarray)로 변환
x_data = np.array(x)
y_data = np.array(y)

#학습률 설정
learning_rate = 0.03

#학습 전체 반복 횟수 설정
epochs = 1001 #0~1000번 총 1001번 반복

# 초기 기울기 (랜덤, 아무 w,b)
w = 0
b = 0

#학습 시작
for i in range(epochs):
    #예측값 y = wx + b
    y_pred = w * x_data + b
    #오차 = 실제값 - 예측값
    error = y_data - y_pred
    #가중치 변화에 따른 손실 변화율 gradient
    #현재 가중치와 loss의 관계에서의 기울기 gradient를 구하기 위해 미분
    w_diff = 2/len(x_data) * sum(((w * x_data + b) - y_data) * x_data)
    #현재 편향과 loss의 관계에서 기울기 gradient를 구하기 위한 미분
    b_diff = 2/len(x_data) * sum((w * x_data + b) - y_data)
    w = w - learning_rate * w_diff
    b = b - learning_rate * b_diff
    if i % 100 == 0:
        print("epoch=%.f, 기울기=%.04f, 절편=%.04f" % (i, w, b))

#예측값
y_pred = w * x_data + b
#그래프 표시
plt.scatter(x, y)
plt.plot([min(x_data), max(x_data)], [min(y_pred), max(y_pred)], color='red')
# 그래프 그리기 위한 좌표값을 파라미터로 받는다
# plt.plot(x축_데이터, y축_데이터)
plt.show()
```
